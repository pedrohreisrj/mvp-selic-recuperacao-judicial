## Setup do ambiente
Criação do catálogo do projeto (`mvp_juros_rj`) e dos três schemas que
representam a arquitetura medalhão (`bronze`, `silver`, `gold`), além de
um Volume (`bronze.arquivos_brutos`) para armazenar arquivos brutos que
não vêm de API (como o export da Serasa Experian).

```python
%sql
CREATE CATALOG IF NOT EXISTS mvp_juros_rj;
CREATE SCHEMA IF NOT EXISTS mvp_juros_rj.bronze;
CREATE SCHEMA IF NOT EXISTS mvp_juros_rj.silver;
CREATE SCHEMA IF NOT EXISTS mvp_juros_rj.gold;
```

```python
%sql
CREATE VOLUME IF NOT EXISTS mvp_juros_rj.bronze.arquivos_brutos;
```
